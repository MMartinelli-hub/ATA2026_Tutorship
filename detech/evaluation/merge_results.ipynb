```python
import pandas as pd

evaluation_results = pd.read_csv('subtask_a_evaluation_results.csv')
baseline_results = pd.read_csv('../baseline/subtask_a_baseline_results.csv')

merged_results = pd.concat([evaluation_results, baseline_results], ignore_index=True)

merged_results = merged_results.sort_values(by='micro_f1', ascending=False).reset_index(drop=True)
merged_results.to_csv('subtask_a_merged_results_sort_by_micro_f1.csv', index=False)

merged_results = merged_results.sort_values(by='type_f1', ascending=False).reset_index(drop=True)
merged_results.to_csv('subtask_a_merged_results_sort_by_type_f1.csv', index=False)
```
